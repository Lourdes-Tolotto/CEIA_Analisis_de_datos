## Tipos de variables

En este bloque se identifican los tipos de datos presentes en el dataset.  
Esto permite distinguir entre variables numéricas, categóricas, booleanas o de fecha, lo cual es importante para decidir qué análisis y visualizaciones aplicar.

```python
# Mostrar todas las filas
pd.set_option("display.max_rows", None)

print("=== Tipos de datos de cada variable ===")
display(df.dtypes.to_frame(name="Tipo"))

print("\n=== Conteo de tipos de variables ===")
display(df.dtypes.value_counts())

print("\n=== Variables numéricas ===")
display(df.select_dtypes(include=["number"]).columns)

print("\n=== Variables categóricas ===")
display(df.select_dtypes(include=["object", "category", "string"]).columns)

print("\n=== Variables booleanas ===")
display(df.select_dtypes(include=["bool"]).columns)

print("\n=== Variables de fecha ===")
display(df.select_dtypes(include=["datetime"]).columns)
```

## Calidad de datos: duplicados, valores nulos y outliers

Se analizan posibles problemas en la calidad del dataset.
Se identifican registros duplicados, variables con valores faltantes y posibles outliers en variables numéricas mediante el método del rango intercuartílico (IQR).
Solo se muestran las variables que presentan nulos o outliers, lo que permite enfocarse en inconsistencias que podrían afectar el análisis posterior.

```python
# Duplicados
print("=== Registros duplicados ===")
duplicados = df.duplicated().sum()
print("Cantidad de filas duplicadas:", duplicados)

# Valores nulos
print("\n=== Valores nulos por variable (solo > 0) ===")
nulos = df.isnull().sum()
nulos_filtrados = nulos[nulos > 0]  # solo columnas con nulos
display(nulos_filtrados)

# Porcentaje de nulos
print("\n=== Porcentaje de valores nulos (solo > 0%) ===")
porcentaje_nulos = (df.isnull().sum() / len(df)) * 100
porcentaje_nulos_filtrado = porcentaje_nulos[porcentaje_nulos > 0]  # solo > 0%
display(porcentaje_nulos_filtrado)

# Outliers con IQR
print("\n=== Outliers en variables numéricas (solo columnas con outliers) ===")
df_num = df.select_dtypes(include="number")
Q1 = df_num.quantile(0.25)
Q3 = df_num.quantile(0.75)
IQR = Q3 - Q1

outliers = ((df_num < (Q1 - 1.5 * IQR)) | (df_num > (Q3 + 1.5 * IQR)))
outliers_count = outliers.sum()

# Filtrar solo columnas con outliers
outliers_filtrados = outliers_count[outliers_count > 0]
display(outliers_filtrados)
```

```python
# Algunos graficos basicos con info macro del dataset

# Config de la figura: 2 filas, 3 columnas
fig = plt.figure(figsize=(18, 10))

#---Layout---

# Fila superior: 3 graficos
ax1 = plt.subplot2grid((2, 3), (0, 0)) 
ax2 = plt.subplot2grid((2, 3), (0, 1)) 

#---Graficos---

# Eventos por region

region_counts = df['region'].value_counts()
ax1.pie(region_counts, labels=region_counts.index, autopct='%1.1f%%', 
        startangle=140, colors=sns.color_palette('pastel'), textprops={'fontsize': 15})
ax1.set_title('Eventos por Región', fontsize=17)

# Eventos por decada
df['decada'] = (df['year'] // 10) * 10
decade_counts = df.groupby('decada').size()
decade_counts.plot(kind='bar', ax=ax2, color='#ffafcc')
ax2.set_title('Eventos por Decada')
ax2.set_xlabel('Periodo')
ax2.set_ylabel('Cantidad de Eventos')
ax2.tick_params(axis='x', rotation=0)


plt.tight_layout()
plt.show()
```

```python

Q1 = df['best'].quantile(0.25)
Q3 = df['best'].quantile(0.75)
IQR = Q3 - Q1
limite_superior = Q3 + 1.5 * IQR

# Filtramos y contamos
outliers = df[df['best'] > limite_superior]
num_outliers = len(outliers)
max_val = df['best'].max()
filas = len(df)

# --- 1. Generación del Gráfico ---
plt.figure(figsize=(10, 3))

# Dibujamos el boxplot base (delgado)
ax = sns.boxplot(data=df, x='best', color='orange', width=0.3, 
                 flierprops={'marker': '', 'alpha':0}) # Ocultamos puntos duplicados

# Superponemos los puntos reales (Jitter) para ver la densidad
sns.stripplot(data=df, x='best', color='red', alpha=0.1, jitter=0.2, size=2)

# Escala Logarítmica (Vital para distribuciones con alta asimetría) 
ax.set_xscale('log')

# Títulos y etiquetas
plt.title('Distribución de Fatalidades (Outliers)', fontsize=19, pad=20)
plt.xlabel('Cantidad de Fatalidades (Escala Log)', fontsize=16)

# Cuadro de info técnica
info_text = (f'Total Outliers: {num_outliers}\n'
             f'Max Fatalidades: {max_val}\n'
             f'% del Dataset: {((num_outliers/filas)*100):.2f}%')

ax.text(0.95, 0.70, info_text, transform=ax.transAxes, 
        ha='right', va='top', fontsize=13, fontweight='bold',
        bbox=dict(facecolor='white', edgecolor='red', boxstyle='round,pad=0.5', alpha=0.9))

plt.grid(True, which="both", ls="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# 1. Definimos los outliers de 'best' (Corregido)
Q1 = df['best'].quantile(0.25)
Q3 = df['best'].quantile(0.75)
IQR = Q3 - Q1

# Como Q3 e IQR ya son números, no necesitan ['best']
limite_superior = Q3 + 1.5 * IQR 
outliers_best = df[df['best'] > limite_superior]

# 2. Filtramos por Middle East
outliers_middle_east = outliers_best[outliers_best['region'] == 'Middle East']

# 3. Calculamos la coincidencia
cantidad_total_outliers = len(outliers_best)
cantidad_me_outliers = len(outliers_middle_east)

porcentaje = (cantidad_me_outliers / cantidad_total_outliers) * 100

print(f"Total de outliers en el mundo: {cantidad_total_outliers}")
print(f"Outliers en Middle East: {cantidad_me_outliers}")
print(f"Porcentaje de outliers que son de Middle East: {porcentaje:.2f}%")
```
